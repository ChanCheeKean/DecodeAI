## Why this topic

Frontier agent behaviour (long tool chains, recovering from failed calls, knowing when to stop)
is now mostly trained, not prompted. *GRPO with RLVR* already covers the GRPO objective, verifiable
rewards and a single-turn tool-call reward; this notebook only shows what **changes** when the
episode has many turns:

| Single-turn RLVR | Multi-turn agentic RL |
|---|---|
| One completion per prompt | Trajectory interleaving model tokens and environment observations |
| Loss on every generated token | Loss masked to model tokens; tool outputs are context, not targets |
| Reward on the final answer | Sparse end-of-episode reward; turn-level credit assignment |
| Batched generation | Rollouts block on tools; async / decoupled tool servers |
| Fixed prompt set | Environments and task synthesis are the bottleneck |
